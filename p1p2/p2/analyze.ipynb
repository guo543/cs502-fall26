```python
import pandas as pd
```

```python
with open("results.txt") as f:
    resultsraw = f.read()
```

```python
print(resultsraw)
```

```python
results = resultsraw.split("####################")
```

```python
names = []
syntaxpoints = []
typepoints = []
feedbacks = []

for u in results:
    try:
        ulist = u.strip().split("\n")
        name = ulist[0][1:]
        synpoint = (50-int(ulist[-1].split()[3]))
        typpoint = (50-int(ulist[-1].split()[6]))
        feedback = "\n".join(ulist[1:-1])
        names.append(name)
        syntaxpoints.append(synpoint)
        typepoints.append(typpoint)
        feedbacks.append(feedback)
    except:
        continue
```

```python
marks = pd.DataFrame({"name": names, "syntaxpoints": syntaxpoints, "typepoints": typepoints, "feedback": feedbacks})
```

```python
marks
```

```python
with open("submitdates.txt") as f:
    submitdates = f.read()
```

```python
submitdateslist = submitdates.strip().split("\n")
```

```python
dnames = []
dates = []
times = []

for sub in submitdateslist:
    subl = sub.split()
    name = subl[0][:-2]
    date = subl[4]
    time = subl[5][:8]
    dnames.append(name)
    dates.append(date)
    times.append(time)
  

subdates = pd.DataFrame({"name": dnames, "date": dates, "times": times})
```

```python
merged = subdates.merge(marks, on='name', how='outer')
```

```python
merged.to_csv("grades.csv", index=False)
```
